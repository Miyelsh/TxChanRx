```python
seed = 1

start_time = time.clock_gettime(0)

TxChanRx.test_snr_sweep(random_seed=seed,
                        bits_per_symbol=bits_per_symbol,
                        symbol_power=symbol_power,
                        num_symbols=num_symbols,
                        num_chan_filter_coefs=num_chan_filter_coefs,
                        num_eq_filter_coefs=num_eq_filter_coefs,
                        chan_filter_noise_power=chan_filter_noise_power,
                        snr_db_sweep=snr_db_sweep,
                        test_dpae=True,
                        dpae_mu=dpae_mu)

end_time = time.clock_gettime(0)

print(f"Run time = {int(end_time-start_time)} seconds")
```
